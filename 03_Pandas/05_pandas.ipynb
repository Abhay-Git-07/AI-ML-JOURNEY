```python
import pandas as pd
```

## Data Cleaning :-
#### - It refers to the process of fixing/removing incomplete, incorrect, or irrelevent data.
#### - ~75% of work done with Pandas is Data Cleaning.

```python
df = pd.read_csv("data.csv")
print(f"Original Data Frame:-\n{df}")

# Dropping Irrelevant Columnns
df = df.drop(columns = ["Legendary", "No"])
print(f"After Data Cleaning:-\n{df}")
```

```python
# Dropping Missing Rows

df = df.dropna(subset = ["Type2"])
print(f"Pokemons with Not null Type2:-\n{df.to_string()}")

# TO DROP/DELETE DUPLICATE VALUES WE USE "df.drop_duplicate()" THIS DELETES ALL DUPLICATE VALUES FROM ALL ROWS
```

```python
# Replacing missing values instead of dropping

df = pd.read_csv("data.csv")
df = df.fillna({"Type2" : "None"})
print(f"Replaced Values:-\n{df.to_string()}")
```

```python
# Fixing inconsistent values

df["Type1"] = df["Type1"].replace({"Grass" : "GRASS", "Fire" : "FIRE"})
print(f"Replaced Values:-\n{df.to_string()}")
```

```python
# Standardize text

df["Name"] = df["Name"].str.upper()
print(f"Names in uppercase letters:-\n{df.to_string()}")
```

```python
# Fixing datatypes

df["Legendary"] = df["Legendary"].astype(bool)
print(f"Integer to Boolean datatype:-\n{df.to_string()}")
```
