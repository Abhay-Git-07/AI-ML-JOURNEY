```python
import pandas as pd
```

## # Data Frame

#### - A tabular data structure with rows and columns.(2 Dimensional).
#### Similar to an excel Spreadsheet.

```python
data = {
    "Name" : ["Kartik","Abhay","Aman"] , 
    "Age" : [17,18,19]
        }

df = pd.DataFrame(data)
print(df)
print("****************************")
df = pd.DataFrame(data, index=["Employee 1", "Employee 2", "Employee 3"])
print(df)
```

```python
print(df.loc["Employee 1"])

print(df.iloc[1])
```

### Adding  a new column/row

```python
# Adding column
df["Job"] = ["Intern", "Developer", "Manager"]
print(df)

print("*" *40)

# Adding rows
new_rows = pd.DataFrame(
    [
    {"Name":"Akshay", "Age":20, "Job":"Cashier"},
    {"Name":"Pratiyank", "Age":18, "Job":"Engineer"}
    ],
    index=["Employee 4","Employee 5"]
    )
df = pd.concat([df,new_rows])
print(df)
```
