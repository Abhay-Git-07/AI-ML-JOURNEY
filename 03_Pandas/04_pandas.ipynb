```python
import pandas as pd
```

## Aggregate Functions
#### :- Reduces a set of values into a single summary value.
####  - Used to summarise and analyse data.
####  - Often used with groupby() function.

```python
df = pd.read_csv("data.csv")

# Finding mean of all numeric functions

print(df.mean(numeric_only=True))
```

```python
# Finding Sum of all numeric values

print(df.sum(numeric_only=True))
```

```python
# Minimum values 
print(df.min(numeric_only=True))

# Maximum Values
print(df.max(numeric_only=True))
```

```python
# For single column

# Mean
print(f"Mean Height:-\n{df["Height"].mean()}")

# Sum
print(f"Total Legendary Pokemons:-\n{df["Legendary"].sum()}")

# Minimum value
print(f"Minimum Weight:-\n{df["Weight"].min()}")

# Maximum Value
print(f"Maximum Weight:-\n{df["Weight"].max()}")

# Total
print(f"Total pokemons with Type 2:-\n{df["Type2"].count()}")
```

```python
# Using groupby() function

group = df.groupby("Type1")  # Grouping by type of each pokemon

# Mean height of each type of pokemon
print(f"Mean Height:-\n{group["Height"].mean()}")  

# Max height of each type 
print(f"Max Height:-\n{group["Height"].max()}")

# Number of pokemons of each type
print(f"Total Height of Pokemons of each type:-\n{group["Type1"].count()}")
```
