```python
import pandas as pd
```

## # Series

#### - A Pandas 1-D labeled array that can hold any datatype.
####          Think of it as a single column in a spreadsheet(1-Dimensional) 


```python
data = [100 , 110 , 120]

series = pd.Series(data) # Series is a constructor not a function
print(series)

series = pd.Series(data, index = ["a","b","c"])  # Passing index ourselves
print(series)
```

```python
print(series.loc["a"])  # We can access particular element with loc property

series.loc["c"] = 200 # Changing the particular value

print(series)

print(series.iloc[0])  # To search value by index
```

### • Filtering the values

```python
data = [100 , 110 , 120 , 130 , 140]

series = pd.Series(data, index = ["a","b","c","d","e"])

print(series >= 120)

print(series[series >= 120])
```

```python
calories = {"Day 1" : 1750 , "Day 2" : 2100, "Day 3" : 1700}

series = pd.Series(calories)
print(series)

series.loc["Day 3"] += 500
print(series)

print(series[series >= 2000])
```
