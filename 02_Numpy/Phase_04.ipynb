```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
array_1 = np.array([[1,2,3],[4,5,6]])
array_2 = np.random.rand(3,3)
array_3 = np.zeros((4,4))
```

```python
# Saving all these array

np.save('array_1.npy', array_1)
np.save('array_2.npy', array_2)
np.save('array_3.npy', array_3)
```

```python
loaded_array_1 = np.load('array_1.npy')
print("Loaded Array 1 : ", loaded_array_1)
loaded_array_2 = np.load('array_2.npy')
print("Loaded Array 2 : ", loaded_array_2)
loaded_array_3 = np.load('array_3.npy')
print("Loaded Array 3 : ", loaded_array_3)
```

```python
try:
    logo = np.load('numpy_logo.npy')
    
    # Display the logo

    plt.figure(figsize=(10,5))
    plt.subplot(121)
    plt.imshow(logo)
    plt.title("Numpy Logo")
    plt.grid(False)

    # Dark mode of this logo

    dark_logo = 1 - logo
    plt.subplot(122)
    plt.imshow(dark_logo)
    plt.title("Numpy Dark Logo")
    plt.grid(False)

except FileNotFoundError:
    print("Numpy logo file not found")
```
