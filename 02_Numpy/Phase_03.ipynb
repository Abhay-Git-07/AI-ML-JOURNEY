```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structure : [restraunt_id , 2021 , 2022, 2023, 2024]

sales_data = np.array([
    [1, 100000, 150000, 170000, 215000], # Pizza king
    [2, 120000, 160000, 180000, 210000], # Shiva Dhaba
    [3, 130000, 150000, 180000, 220000], # Mevar
    [4, 125000, 170000, 190000, 240000], # Grill house
    [5, 160000, 190000, 210000, 230000]  # Moga
])
print("===Zomato Sales Analysis===")
print("\nSales Data Shape :-", sales_data.shape)
print("\nSales of first 3 restraunts :-", sales_data[:3]) # [0:3 bhi likh skte the pr ye prefer krte hain]
print("\nOnly Sales of all restraunts :-", sales_data[: , 1:])
```

```python
# Total sales per year

print(np.sum(sales_data,axis=0))
yearly_sum = np.sum(sales_data[:,1:], axis=0)
print("Yearly sales :",yearly_sum)

# Minimum sales per restraunt

min_sales = np.min(sales_data[:, 1:], axis=1)
print("Minimum sales per restraunt : ",min_sales)

# Maximum sales per year 

max_sales = np.max(sales_data[:, 1:], axis=0)
print("Max sale per year : ",max_sales)

# Avg Sales per restaurant

avg_sales = np.mean(sales_data[:, 1:], axis=1)
print("Average sales of each restraunt : ",avg_sales)
```

```python
# Cumulative Sales of all restraunts

cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print("Cumulative sales : ",cumsum)

# Graph of cumulative sales per year

plt.figure(figsize=(8,6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumulative sales across all restraunt")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
# Vector Operations

vector_1 = np.array([1,2,3,4,5])
vector_2 = np.array([6,7,8,9,10])

# Vector Addition

print("Vector Addition : ",vector_1 + vector_2)

# Vector Multiplication

print("\nVector Multiplication : ",vector_1 * vector_2)

# Dot Product

print("\nDot Product : ", np.dot(vector_1,vector_2))

# Angle between vectors

angle = np.arccos(np.dot(vector_1,vector_2) / (np.linalg.norm(vector_1) * np.linalg.norm(vector_2)))

print("Angle b/w vectors : ",angle)
```

```python
# Vectorisation of a vector

restraunt_types = np.array(["biryani", "pizza" , "burger", "chinese", "cafe"])

# Convert into uppercase 

vectorised_upper = np.vectorize(str.upper)
print("Vectorised Upper : ", vectorised_upper(restraunt_types))

# Monthly Average

monthly_avg = sales_data[:, 1:] / 12
print("Monthly Average : ",monthly_avg)
```
